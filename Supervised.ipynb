```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
X,y_true = make_blobs(n_samples=500, centers=3, cluster_std=0.60, random_state=42)
```

```python
X
```

```python
df = pd.DataFrame(X,columns=['Feature1','Feature2'])
```

```python
df
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
inertia = []
K = range(1,11)
for k in K:
    kmeans = KMeans(n_clusters=k,random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
inertia
```

```python
plt.plot(K,inertia,'-o')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.show()
#
```

```python
Kmeans_final = KMeans(n_clusters=3,random_state=42)
cluster_labels = Kmeans_final.fit_predict(X_scaled)
```

```python
cluster_labels
```

```python
df['cluster'] = cluster_labels
```

```python
sns.scatterplot(data=df,x='Feature1',y='Feature2',hue='cluster')
```

```python
from sklearn.datasets import make_moons
```

```python
X,y_true = make_moons(n_samples=500, noise = 0.05, random_state=42)
```

```python
from sklearn.cluster import KMeans,DBSCAN
```

```python
df = pd.DataFrame(X,columns=['Feature1','Feature2'])
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
Kmeans = KMeans(n_clusters=2,random_state=42)
Kmeans_labels = Kmeans.fit(X_scaled)
```

```python
df['Kmeans_cluster'] = Kmeans_labels.labels_
```

```python
sns.scatterplot(data=df,x='Feature1',y='Feature2',hue='Kmeans_cluster',palette='tab10')
```

```python
dbscan = DBSCAN(eps=0.3,min_samples=5)
dbscan_labels = dbscan.fit_predict(X_scaled)
```

```python
df['dbscan_cluster'] = dbscan_labels
```

```python
sns.scatterplot(data=df,x='Feature1',y='Feature2',hue='dbscan_cluster',palette='tab10')
```
